```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

## Q1.

**Problem Statement:**

Ask the user to enter a value (e.g., M) for a mathematical function.
Generate x values from -10 to 10 using np.linspace().
Compute y values for:
- y= M⋅x2
- y=M⋅sin(x)

Plot both functions in a single figure:
- Use different colors and line styles.
- Add legend, grid, and title.

```python
M = float(input("Enter a value for M: "))
x = np.linspace(-10, 10, 400)
y1 = M * x**2
y2 = M * np.sin(x)
plt.figure(figsize=(8, 5))
plt.plot(x, y1, color='blue', linestyle='-', label=f'y = {M}·x²')
plt.plot(x, y2, color='red', linestyle='--', label=f'y = {M}·sin(x)')

plt.title('Plot of y = M·x² and y = M·sin(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```

## Q.2

**Problem Statement:**

Create a dataset of five subjects and scores.
Convert it into a Pandas DataFrame.
Plot the scores using a Seaborn bar plot with:
- Different colors for each bar.
- Annotations on top of each bar.
- Title, axis labels, and grid.

```python
data = {
    'Subject': ['Math', 'Physics', 'Chemistry', 'IT', 'English'],
    'Score': [100, 82, 86, 93, 95]
}
df = pd.DataFrame(data)

plt.figure(figsize=(8, 5))
ax = sns.barplot(x='Subject', y='Score', hue='Subject', data=df, palette='viridis', legend=False)
for i, score in enumerate(df['Score']):
    ax.text(i, score + 1, str(score), ha='center', va='bottom', fontweight='bold')
plt.title('Subject-wise Scores')
plt.xlabel('Subject')
plt.ylabel('Score')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.ylim(0, 100)
plt.show()
```

## Q3.

**Problem Statement:**

Write a Python script to select your roll number as seed for NumPy and
then generate a dataset of 50 values using np.random.randn().
Create a 2x2 subplot layout displaying:
- Line plot showing cumulative sum.
- Scatter plot with random noise.

Customize with titles, axis labels, and grids.

```python
roll_number = 1024170164
np.random.seed(roll_number)

data = np.random.randn(50)
cumulative_sum = np.cumsum(data)
noise = np.random.randn(50)
fig, axs = plt.subplots(2, 2, figsize=(10, 8))
# LINE PLOT OF CUMMULATIVE SUM
axs[0, 0].plot(cumulative_sum, color='blue', linestyle='-')
axs[0, 0].set_title('Cumulative Sum (Line Plot)')
axs[0, 0].set_xlabel('Index')
axs[0, 0].set_ylabel('Cumulative Sum')
axs[0, 0].grid(True)
#SCATTER PLOT OF RANDOM NOISE
axs[0, 1].scatter(range(50), noise, color='red')
axs[0, 1].set_title('Random Noise (Scatter Plot)')
axs[0, 1].set_xlabel('Index')
axs[0, 1].set_ylabel('Noise Value')
axs[0, 1].grid(True)
#LINE PLOT OF RAW RANDOM DATA
axs[1, 0].plot(data, color='green', linestyle='--')
axs[1, 0].set_title('Raw Random Data (Line Plot)')
axs[1, 0].set_xlabel('Index')
axs[1, 0].set_ylabel('Value')
axs[1, 0].grid(True)
#SCATTER PLOT OF DATA VS NOISE
axs[1, 1].scatter(data, noise, color='purple')
axs[1, 1].set_title('Data vs Noise (Scatter Plot)')
axs[1, 1].set_xlabel('Data')
axs[1, 1].set_ylabel('Noise')
axs[1, 1].grid(True)

plt.tight_layout()
plt.show()
```

## Q.4

**Problem Statement:**

Download the dataset from the link below and apply the seaborn library to perform the following tasks.

Link: https://github.com/AnjulaMehto/MCA/blob/main/company_sales_data.csv

**Requirements:**

1. Read Total profit of all months and show it using a line plot.
2. Read all product sales data and show it using a multiline plot.
3. Plot bar chart for all the features/attributes.

```python
sns.set_style("whitegrid")

df = pd.read_csv('company_sales_data.csv')
print(df)

# 1. Total profit of all months - Line plot
plt.figure(figsize=(8, 5))
sns.lineplot(x='month_number', y='total_profit', data=df, marker='o', color='green')
plt.title('Total Profit per Month')
plt.xlabel('Month Number')
plt.ylabel('Total Profit')
plt.grid(True)
plt.show()

# 2. All product sales data - Multiline plot
products = ['facecream', 'facewash', 'toothpaste', 'bathingsoap', 'shampoo', 'moisturizer']
plt.figure(figsize=(9, 6))
for product in products:
    sns.lineplot(x='month_number', y=product, data=df, marker='o', label=product)
plt.title('Product Sales per Month')
plt.xlabel('Month Number')
plt.ylabel('Units Sold')
plt.legend(title='Product')
plt.grid(True)
plt.show()

# 3. Bar chart for all features/attributes
features = df.columns.drop('month_number')
plt.figure(figsize=(10, 6))
df[features].sum().plot(kind='bar', color=sns.color_palette('viridis', len(features)))
plt.title('Total of Each Feature/Attribute (All Months Combined)')
plt.xlabel('Feature')
plt.ylabel('Total Value')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```
